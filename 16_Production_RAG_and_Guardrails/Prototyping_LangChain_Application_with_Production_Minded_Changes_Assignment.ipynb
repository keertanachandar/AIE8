#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

    
> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer

- **Memory vs Disk caching trade-offs**
    - In-Memory: Useful when you need extremely fast access to small sets of data. Examples include session data, or any data that is frequently accessed. It won't be good for data that needs to be persistant across sessions/restarts etc., or are large.
    - Disk: Useful when you need to store/access large static assets like images/videos. This isn't the best when you are frequently updating the data or need very quick access.
- **Cache invalidation strategies** 
    - TTL: Useful when there are predicted updates in the data, for example weather updates or stock updates. Not useful when you need real time data or there are unpredictable data updates. 
    - Update cache every time data is updated: This means updating based on what is the most used. Most useful when data consistency is a priority. This won't be very useful if you need to write to the cache very often or if the updates are hard to 
    - Optimistic updates: Update the cache first and later update the database. Useful when you need instant UI feedback and can't wait for the server. Examples include chat updates or video games. Not useful for metrics dashboards or leaderboards where the information should be the same for everyone.
    - Lazy Loading: Only put into the cache when someone asks for it. Useful in web development when the first load is slow but all subsequent loads load immediately since it is stored. Bad for data that you know all users will need.
    - Manual: This is useful if you don't need to update the cache very often, or if there is a clear event that calls for the cache to be updated. It won't be useful when you need to make frequent updates.
- **Concurrent access patterns**
Problem: Cache invalidates at 3am, and at 3:00:01 1k people request the same data. All 1000 check the cache but miss so then make a database call which overloads the system.
    - Read Heavy: This is the most useful - where users are reading the majority of the time and not writing to the cache very often. In the problem above, the first person will get the cache miss and then call the database. The rest will wait for the cache to populate.
    - Write Heavy: This is when there are multiple people writing at the same time, for example like live comments or a twitter thread. Instead of using a cache for each user, you can can have aggregate data grouped together like "popular".
- **Cache size management**
    - Least Recently Used: Useful for most cases and access patterns are somewhat stable. Not useful when there are random access patterns. 
    - Least Frequently Used: Useful for top information that stays somewhat stable. Not useful for fast trends or bursts of data.
    - Fixed Size: Useful for a data that you know will always stay a certain size. Not useful when the data sets are variable.
- **Cold start scenarios**
Problem: Your cache is completely empty so you're starting from scratch.
    - Cache warming: Don't let users access the cache until there is data in there again. Useful for information you know most users will use.
    - Lazy loading: Only cache the information that people have requested. If everyone hits the cache at the same time, there may be some loadbearing issues. It also would take longer to fill up the cache.
    - Hybrid: Load the cache with a portion of pre-loaded data. The rest can be lazy loaded.
    


##### 🏗️ Activity #1: Cache Performance Testing

Create a simple experiment that tests our production caching system:

1. **Test embedding cache performance**: Try embedding the same text multiple times
2. **Test LLM cache performance**: Ask the same question multiple times  
3. **Measure cache hit rates**: Compare first call vs subsequent calls

````python
### Cache Performance Testing Experiment ###

import time
from typing import List, Tuple

print("🧪 CACHE PERFORMANCE TESTING EXPERIMENT")
print("=" * 60)

# ============================================================================
# Test 1: Embedding Cache Performance
# ============================================================================
print("\n📊 TEST 1: EMBEDDING CACHE PERFORMANCE")
print("-" * 60)

# Test texts of varying lengths
test_texts = [
    "What are the interest rates for Direct Subsidized Loans?",
    "The Direct Loan Program provides federal student loans to help students pay for college. These loans include Direct Subsidized Loans, Direct Unsubsidized Loans, Direct PLUS Loans, and Direct Consolidation Loans.",
    "Understanding repayment options is crucial for borrowers. Income-driven repayment plans, standard repayment, and graduated repayment are all available options that borrowers should consider based on their financial situation."
]

embedding_times = {"first_call": [], "cache_hit": []}

for i, text in enumerate(test_texts, 1):
    print(f"\n  Test {i}: Embedding text ({len(text)} chars)")
    
    # First call - will compute and cache
    start = time.time()
    embeddings_1 = rag_chain.cached_embeddings.get_embeddings().embed_query(text)
    first_time = time.time() - start
    embedding_times["first_call"].append(first_time)
    print(f"    ⏱️  First call (cache miss):  {first_time:.4f}s")
    
    # Second call - should use cache
    start = time.time()
    embeddings_2 = rag_chain.cached_embeddings.get_embeddings().embed_query(text)
    cache_time = time.time() - start
    embedding_times["cache_hit"].append(cache_time)
    print(f"    ⚡ Second call (cache hit):  {cache_time:.4f}s")
    
    speedup = first_time / cache_time if cache_time > 0 else float('inf')
    print(f"    🚀 Speedup: {speedup:.1f}x faster")
    
    # Verify embeddings are identical (compare as lists)
    import numpy as np
    if np.allclose(embeddings_1, embeddings_2, rtol=1e-9):
        print(f"    ✅ Embeddings match (cached correctly)")
    else:
        print(f"    ⚠️  Warning: Embeddings differ slightly")

# Calculate average speedup for embeddings
avg_first = sum(embedding_times["first_call"]) / len(embedding_times["first_call"])
avg_cache = sum(embedding_times["cache_hit"]) / len(embedding_times["cache_hit"])
avg_speedup = avg_first / avg_cache if avg_cache > 0 else float('inf')

print(f"\n  📈 EMBEDDING CACHE SUMMARY:")
print(f"     Average first call:  {avg_first:.4f}s")
print(f"     Average cache hit:   {avg_cache:.4f}s")
print(f"     Average speedup:     {avg_speedup:.1f}x")

# ============================================================================
# Test 2: LLM Cache Performance
# ============================================================================
print("\n\n📊 TEST 2: LLM CACHE PERFORMANCE")
print("-" * 60)

# Test questions
test_questions = [
    "What types of loans are available in the Direct Loan Program?",
    "How do I apply for a Direct Loan?",
    "What are the repayment options for Direct Loans?"
]

llm_times = {"first_call": [], "cache_hit": []}

for i, question in enumerate(test_questions, 1):
    print(f"\n  Test {i}: {question}")
    
    # First call - will hit OpenAI and cache
    start = time.time()
    response_1 = rag_chain.invoke(question)
    first_time = time.time() - start
    llm_times["first_call"].append(first_time)
    print(f"    ⏱️  First call (cache miss):  {first_time:.4f}s")
    print(f"    📝 Response preview: {response_1.content[:100]}...")
    
    # Second call - should use cache
    start = time.time()
    response_2 = rag_chain.invoke(question)
    cache_time = time.time() - start
    llm_times["cache_hit"].append(cache_time)
    print(f"    ⚡ Second call (cache hit):  {cache_time:.4f}s")
    
    speedup = first_time / cache_time if cache_time > 0 else float('inf')
    print(f"    🚀 Speedup: {speedup:.1f}x faster")
    
    # Verify responses are identical (cached)
    if response_1.content == response_2.content:
        print(f"    ✅ Responses match (cached correctly)")
    else:
        print(f"    ⚠️  Warning: Responses differ (cache may not be working)")

# Calculate average speedup for LLM
avg_first_llm = sum(llm_times["first_call"]) / len(llm_times["first_call"])
avg_cache_llm = sum(llm_times["cache_hit"]) / len(llm_times["cache_hit"])
avg_speedup_llm = avg_first_llm / avg_cache_llm if avg_cache_llm > 0 else float('inf')

print(f"\n  📈 LLM CACHE SUMMARY:")
print(f"     Average first call:  {avg_first_llm:.4f}s")
print(f"     Average cache hit:   {avg_cache_llm:.4f}s")
print(f"     Average speedup:     {avg_speedup_llm:.1f}x")

# ============================================================================
# Test 3: Cache Hit Rate Analysis
# ============================================================================
print("\n\n📊 TEST 3: CACHE HIT RATE ANALYSIS")
print("-" * 60)

# Simulate a workload with repeated queries
workload_questions = [
    "What types of loans are available?",
    "How do I repay my loan?",
    "What types of loans are available?",  # Repeat
    "What are the interest rates?",
    "How do I repay my loan?",  # Repeat
    "What types of loans are available?",  # Repeat
    "Can I consolidate my loans?",
    "What are the interest rates?",  # Repeat
]

print(f"\n  Simulating workload with {len(workload_questions)} queries...")
print(f"  (Including repeated queries to test cache effectiveness)\n")

workload_times = []
cache_hits = 0
cache_misses = 0

for i, question in enumerate(workload_questions, 1):
    start = time.time()
    response = rag_chain.invoke(question)
    elapsed = time.time() - start
    workload_times.append(elapsed)
    
    # Heuristic: if response is very fast, it was likely a cache hit
    if elapsed < 0.5:  # Threshold for cache hit
        cache_hits += 1
        status = "✅ HIT "
    else:
        cache_misses += 1
        status = "❌ MISS"
    
    print(f"  Query {i}: {status} - {elapsed:.4f}s - {question[:50]}...")

# Calculate cache hit rate
total_queries = len(workload_questions)
hit_rate = (cache_hits / total_queries) * 100 if total_queries > 0 else 0

print(f"\n  📈 CACHE HIT RATE ANALYSIS:")
print(f"     Total queries:       {total_queries}")
print(f"     Cache hits:          {cache_hits}")
print(f"     Cache misses:        {cache_misses}")
print(f"     Hit rate:            {hit_rate:.1f}%")
print(f"     Avg response time:   {sum(workload_times)/len(workload_times):.4f}s")

# ============================================================================
# Overall Summary
# ============================================================================
print("\n\n" + "=" * 60)
print("🎯 OVERALL CACHE PERFORMANCE SUMMARY")
print("=" * 60)

print(f"\n  📊 Embedding Cache:")
print(f"     • Speedup: {avg_speedup:.1f}x faster")
print(f"     • Time saved per query: {(avg_first - avg_cache):.4f}s")
print(f"     • Efficiency gain: {((1 - avg_cache/avg_first) * 100):.1f}%")

print(f"\n  💬 LLM Cache:")
print(f"     • Speedup: {avg_speedup_llm:.1f}x faster")
print(f"     • Time saved per query: {(avg_first_llm - avg_cache_llm):.4f}s")
print(f"     • Efficiency gain: {((1 - avg_cache_llm/avg_first_llm) * 100):.1f}%")

print(f"\n  🎯 Cache Hit Rate:")
print(f"     • Hit rate: {hit_rate:.1f}%")
print(f"     • Effective for repeated queries: {'✅ YES' if hit_rate > 50 else '⚠️  NEEDS IMPROVEMENT'}")

print(f"\n  💰 Production Impact:")
total_time_without_cache = sum(workload_times) if hit_rate == 0 else (sum(workload_times) / hit_rate) * 100
time_saved = total_time_without_cache - sum(workload_times)
print(f"     • Estimated time saved: {time_saved:.2f}s for {total_queries} queries")
print(f"     • Cost reduction: ~{hit_rate:.1f}% (no redundant API calls)")
print(f"     • Scalability: Much better handling of concurrent requests")

print("\n" + "=" * 60)
print("✅ Cache performance testing complete!")
print("=" * 60)
````

Compare the Simple Agent vs Helpfulness Agent architectures:

   - Simple Agent advantages/disadvantages
   - Helpfulness Agent advantages/disadvantages

2. **Production Considerations:**
   - How does the helpfulness check affect latency?
   - What are the cost implications of iterative refinement?
   - How would you monitor agent performance in production?

3. **Scalability Questions:**
   - How would these agents perform under high concurrent load?
   - What caching strategies work best for each agent type?

##### ✅ Answer

1. **When would you choose each agent type?**
   - Simple Agent: Does you request and that's it.
      - advantages: Fast and cheap, no back and forth, and predictable since it will always do the same thing.
      - disadvantages: Follows instructions blindly and no quality control. Also can't learn from it's behavior.
   - Helpfulness Agent: Checks if the result is helpful, and if not will try to improve the results.
      - advantages: Smart, flexible, handles ambiguity, and will return higher quality results.
      - disadvantages: Slower, more expensive, unpredictable how it will respond, and more complex.

2. **Production Considerations:**
   - How does the helpfulness check affect latency?
      - Helpfulness makes at least 1 additional call to check on the helpfulness of the results. There is a max limit of cycles it can go through, but there will always be more calls than the simple agent because there is one extra node. Each call takes some time.
   - What are the cost implications of iterative refinement?
      - By iteratively refining your agents/results you will be making either more calls, or using a more complex llm, either of which will be more expensive. A simple agent can cost ~3x less than the helpful agent because of the lower number of calls. 
   - How would you monitor agent performance in production?
      - Using langgraph you can monitor how each agent performs in terms of latency, cost, avg. number of iterations, and other metrics that you can set up. For example, helpfulness, accuracy, recall, faithfulness, etc.

3. **Scalability Questions:**
   - How would these agents perform under high concurrent load? E.g. 10,000 users hit your site at once. Each request takes 2 seconds.
      - simple: Best case with unlimited servers - all requests are finished in 2 seconds and costs $X.
      - helpful: Requests pile up since there are requests that are now queued up, and servers might run out of memory. And it costs at least $3X.
   - What caching strategies work best for each agent type?
      - simple: Use a key:value approach and use a longer TTL and lazy load the cache. This works since 1 input means the same output every time so the information wouldn't change too often. It is also fast and cheap.
      - helpful: You can do a partial cache where you store part of the results so that any subsequent calls made through the helpfulness agent are already filtered and you're making less api calls.
   - How would you implement rate limiting and circuit breakers?
      - rate limiting: You can only make X requests in Y minutes/hours/days/etc.
      - circuit breakers: If an API keeps failing then shut it off completely.
      Both options help make sure users are not abusing your system, you are controlling the number of calls being made, and not wasting money.

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

- **Layered Defense**: Multiple validation stages for robust protection

### 🛡️ Activity #3 Implementation: Production-Safe Agent with Guardrails

Let's create a simple wrapper around our existing agent that adds guardrail validation!


````python
# Simple Guardrails-Protected Agent Implementation
from langchain_core.messages import HumanMessage
import time

class GuardedAgent:
    """A simple wrapper that adds guardrail validation to our LangGraph agent."""
    
    def __init__(self, agent, input_guards=None, output_guards=None):
        """
        Initialize the guarded agent.
        
        Args:
            agent: The base LangGraph agent
            input_guards: List of guards to validate input
            output_guards: List of guards to validate output
        """
        self.agent = agent
        self.input_guards = input_guards or []
        self.output_guards = output_guards or []
        self.validation_logs = []
    
    def validate_input(self, user_input):
        """Validate user input through all input guards."""
        for guard in self.input_guards:
            try:
                result = guard.validate(user_input)
                self.validation_logs.append({
                    "stage": "input",
                    "guard": guard.__class__.__name__,
                    "status": "passed",
                    "input": user_input[:50]
                })
            except Exception as e:
                self.validation_logs.append({
                    "stage": "input",
                    "guard": guard.__class__.__name__,
                    "status": "failed",
                    "error": str(e)[:100],
                    "input": user_input[:50]
                })
                raise ValueError(f"Input validation failed: {str(e)[:100]}")
        return True
    
    def validate_output(self, output_text):
        """Validate agent output through all output guards."""
        for guard in self.output_guards:
            try:
                result = guard.validate(output_text)
                self.validation_logs.append({
                    "stage": "output",
                    "guard": guard.__class__.__name__,
                    "status": "passed"
                })
            except Exception as e:
                self.validation_logs.append({
                    "stage": "output",
                    "guard": guard.__class__.__name__,
                    "status": "failed",
                    "error": str(e)[:100]
                })
                # For output, we'll return a safe fallback message
                return "I apologize, but I cannot provide that response as it doesn't meet our content guidelines."
        return output_text
    
    def invoke(self, user_input):
        """
        Process user input through the guarded agent pipeline.
        
        Pipeline: Input Validation → Agent Processing → Output Validation
        """
        start_time = time.time()
        
        # Step 1: Validate Input
        try:
            self.validate_input(user_input)
        except ValueError as e:
            return {
                "success": False,
                "error": str(e),
                "stage": "input_validation",
                "time": time.time() - start_time
            }
        
        # Step 2: Process with Agent
        try:
            messages = [HumanMessage(content=user_input)]
            response = self.agent.invoke({"messages": messages})
            agent_output = response["messages"][-1].content
        except Exception as e:
            return {
                "success": False,
                "error": f"Agent error: {str(e)[:100]}",
                "stage": "agent_processing",
                "time": time.time() - start_time
            }
        
        # Step 3: Validate Output
        safe_output = self.validate_output(agent_output)
        
        return {
            "success": True,
            "input": user_input,
            "output": safe_output,
            "time": time.time() - start_time,
            "validation_logs": self.validation_logs.copy()
        }
    
    def get_logs(self):
        """Get all validation logs."""
        return self.validation_logs


# Create our guarded agent
if guardrails_available and simple_agent:
    print("🛡️ Creating Production-Safe Guarded Agent")
    print("=" * 60)
    
    # Set up guards for input validation
    input_guards = [
        competitor_guard,  # Block competitor mentions
        profanity_guard    # Block inappropriate language
    ]
    
    # Set up guards for output validation  
    output_guards = [
        profanity_guard    # Ensure clean responses
    ]
    
    # Create the guarded agent
    guarded_agent = GuardedAgent(
        agent=simple_agent,
        input_guards=input_guards,
        output_guards=output_guards
    )
    
    print("✅ Guarded Agent Created!")
    print(f"   Input Guards: {len(input_guards)}")
    print(f"   - {', '.join([g.__class__.__name__ for g in input_guards])}")
    print(f"   Output Guards: {len(output_guards)}")
    print(f"   - {', '.join([g.__class__.__name__ for g in output_guards])}")
    print()
    print("🔒 Security Pipeline:")
    print("   User Input → Input Guards → Agent → Output Guards → Response")
    
else:
    print("⚠️  Prerequisites not met. Please run earlier cells first:")
    print("   - Cell 39: Check guardrails availability")
    print("   - Cell 27: Create simple_agent")
    print("   - Cell 43: Configure guards")
````
