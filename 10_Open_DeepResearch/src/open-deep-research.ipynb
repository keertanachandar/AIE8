## LangGraph Open Deep Research - Supervisor-Researcher Architecture

In this notebook, we'll explore the **supervisor-researcher delegation architecture** for conducting deep research with LangGraph.

You can visit this repository to see the original application: [Open Deep Research](https://github.com/langchain-ai/open_deep_research)

Let's jump in!

## What We're Building

This implementation uses a **hierarchical delegation pattern** where:

1. **User Clarification** - Optionally asks clarifying questions to understand the research scope
2. **Research Brief Generation** - Transforms user messages into a structured research brief
3. **Supervisor** - A lead researcher that analyzes the brief and delegates research tasks
4. **Parallel Researchers** - Multiple sub-agents that conduct focused research simultaneously
5. **Research Compression** - Each researcher synthesizes their findings
6. **Final Report** - All findings are combined into a comprehensive report

#### ❓ Question 1:

 Explain the interrelationships between the three states.  Why don't we just make a single huge state?

 Agent State - This level basically orchestrates and maintains the converstaion context for all researchers. It also handles all the final responses. This state will persist throughtout the lifecycle.

 Supervisor State - This level basically coordinates all the research that needs to be done, and delegates to the right tools. It is active during the research phase, and then resets. It doesn't know what the Agent level knows.

 Researcher State - This level is basically the specialist. It focuses on one research question at a time and only needs its own conversation and tools, with no context of anything else.

 One state would be incredibly messy and also inefficient. One state means you can't run multiple research questions in parallel. It also means, that the context would get incredibly muddled with irrelevant information, and the final output would not be as comprehensive or faithful. There would also be no clear definition of responsibilites, so one state means that it would do the task of all 3 states above, but at a lower caliber.

## Why This Architecture?

### Advantages of Supervisor-Researcher Delegation

1. **Dynamic Task Decomposition**
   - Unlike section-based approaches with predefined structure, the supervisor can break down research based on the actual question
   - Adapts to different types of research (comparisons, lists, deep dives, etc.)

2. **Parallel Execution**
   - Multiple researchers work simultaneously on different aspects
   - Much faster than sequential section processing
   - Configurable parallelism (1-20 concurrent researchers)

3. **ReAct Pattern for Quality**
   - Researchers use `think_tool` to reflect after each search
   - Prevents excessive searching and improves search quality
   - Natural stopping conditions based on information sufficiency

4. **Flexible Tool Integration**
   - Easy to add MCP tools for specialized research
   - Supports multiple search APIs (Anthropic, Tavily)
   - Each researcher can use different tool combinations

5. **Graceful Token Limit Handling**
   - Compression prevents token overflow
   - Progressive truncation in final report generation
   - Research can scale to arbitrary depths

### Trade-offs

- **Complexity:** More moving parts than section-based approach
- **Cost:** Parallel researchers use more tokens (but faster)
- **Unpredictability:** Research structure emerges dynamically

````python
# Load the PDF document
from pathlib import Path
import PyPDF2

def load_pdf(pdf_path: str) -> str:
    """Load and extract text from PDF."""
    pdf_text = ""
    with open(pdf_path, 'rb') as file:
        pdf_reader = PyPDF2.PdfReader(file)
        for page in pdf_reader.pages:
            pdf_text += page.extract_text() + "\n\n"
    return pdf_text

# Load the PDF about how people use AI
pdf_path = "data/howpeopleuseai.pdf"
pdf_content = load_pdf(pdf_path)

print(f"Loaded PDF with {len(pdf_content)} characters")
print(f"First 500 characters:\n{pdf_content[:500]}...")
````

````python
# Set up the graph with Anthropic configuration
from IPython.display import Markdown, display
import uuid

# Note: deep_researcher is already compiled from the library
# For this demo, we'll use it directly without additional checkpointing
graph = deep_researcher

print("✓ Graph ready for execution")
print("  (Note: The graph is pre-compiled from the library)")
````

### Execute the Research

Now let's run the research! We'll ask the system to analyze the PDF and provide insights about how people use AI.

The workflow will:
1. **Clarify** - Check if the request is clear (may skip if obvious)
2. **Research Brief** - Transform our request into a structured brief
3. **Supervisor** - Plan research strategy and delegate to researchers
4. **Parallel Research** - Multiple researchers gather information simultaneously
5. **Compression** - Each researcher synthesizes their findings
6. **Final Report** - All findings combined into comprehensive report

````python
# Create our research request with PDF context
research_request = f"""
I have a PDF document about how people use AI. Please analyze this document and provide insights about:

1. What are the main findings about how people are using AI?
2. What are the most common use cases?
3. What trends or patterns emerge from the data?

Here's the PDF content:

{pdf_content[:10000]}  # First 10k chars to stay within limits

...[content truncated for context window]
"""

# Execute the graph
async def run_research():
    """Run the research workflow and display results."""
    print("Starting research workflow...\n")
    
    async for event in graph.astream(
        {"messages": [{"role": "user", "content": research_request}]},
        config,
        stream_mode="updates"
    ):
        # Display each step
        for node_name, node_output in event.items():
            print(f"\n{'='*60}")
            print(f"Node: {node_name}")
            print(f"{'='*60}")
            
            if node_name == "clarify_with_user":
                if "messages" in node_output:
                    last_msg = node_output["messages"][-1]
                    print(f"\n{last_msg.content}")
            
            elif node_name == "write_research_brief":
                if "research_brief" in node_output:
                    print(f"\nResearch Brief Generated:")
                    print(f"{node_output['research_brief'][:500]}...")
            
            elif node_name == "supervisor":
                print(f"\nSupervisor planning research strategy...")
                if "supervisor_messages" in node_output:
                    last_msg = node_output["supervisor_messages"][-1]
                    if hasattr(last_msg, 'tool_calls') and last_msg.tool_calls:
                        print(f"Tool calls: {len(last_msg.tool_calls)}")
                        for tc in last_msg.tool_calls:
                            print(f"  - {tc['name']}")
            
            elif node_name == "supervisor_tools":
                print(f"\nExecuting supervisor's tool calls...")
                if "notes" in node_output:
                    print(f"Research notes collected: {len(node_output['notes'])}")
            
            elif node_name == "final_report_generation":
                if "final_report" in node_output:
                    print(f"\n" + "="*60)
                    print("FINAL REPORT GENERATED")
                    print("="*60 + "\n")
                    display(Markdown(node_output["final_report"]))
    
    print("\n" + "="*60)
    print("Research workflow completed!")
    print("="*60)

# Run the research
await run_research()
````

````python
# integrated_research_run.py - Complete integration with your graph

from pathlib import Path
from quick_setup import QuickSetup

def setup_and_run_research():
    """Complete research workflow with graph integration"""
    
    print("\n" + "="*70)
    print("🔬 INTEGRATED RESEARCH WORKFLOW")
    print("="*70)
    
    # ========================================================================
    # STEP 1: Configuration
    # ========================================================================
    print("\n📋 Step 1: Configuration")
    
    config = QuickSetup.user_research(
        model_preset="balanced",  # or "quality" or "economical"
        budget=25.0
    )
    
    config_dict = config.get_config_dict()
    
    # ========================================================================
    # STEP 2: Load Document
    # ========================================================================
    print("\n📄 Step 2: Loading document...")
    
    # Try multiple possible paths
    possible_paths = [
        Path("documents/w34255.pdf"),
        Path("documents/research.pdf"),
        Path("w34255.pdf"),
        Path("research.pdf")
    ]
    
    pdf_content = None
    for doc_path in possible_paths:
        if doc_path.exists():
            print(f"✓ Found document: {doc_path}")
            try:
                # Try PDF reader first
                try:
                    import PyPDF2
                    with open(doc_path, 'rb') as f:
                        pdf_reader = PyPDF2.PdfReader(f)
                        pdf_content = ""
                        for page in pdf_reader.pages:
                            pdf_content += page.extract_text()
                    print(f"✓ Extracted {len(pdf_content)} characters from PDF")
                except ImportError:
                    # Fallback to text reading
                    with open(doc_path, 'r', encoding='utf-8', errors='ignore') as f:
                        pdf_content = f.read()
                    print(f"✓ Read {len(pdf_content)} characters as text")
                break
            except Exception as e:
                print(f"⚠️  Error reading {doc_path}: {e}")
                continue
    
    if not pdf_content:
        print("⚠️  No document found, using sample content")
        pdf_content = """
        NBER Working Paper: How People Use ChatGPT
        
        Sample findings for testing:
        - 65% of users employ ChatGPT for work-related tasks
        - 45% use it for creative writing and content generation
        - Average session length: 15 minutes
        - Most common use case: Content generation (78%)
        - Education use: 42% of users
        - Code assistance: 38% of users
        
        Demographics:
        - Age 25-34: 35% of users
        - Age 35-44: 28% of users
        - Professional workers: 62%
        
        Trends:
        - Usage increased 150% from 2023 to 2024
        - Mobile usage grew by 80%
        - API usage doubled in 2024
        """
    
    # ========================================================================
    # STEP 3: Create Research Query
    # ========================================================================
    print("\n❓ Step 3: Creating research query...")
    
    research_query = """
    Analyze this document about ChatGPT usage patterns and provide comprehensive insights:
    
    1. **Main Findings** - What are the key findings about how people use AI?
       - Extract all usage statistics and percentages
       - Identify demographic patterns
       - Note adoption trends over time
       - Use the 'extract_statistics' and 'extract_key_findings' tools
    
    2. **Common Use Cases** - What are the most frequent applications?
       - Categorize use cases by domain (work, education, creative, personal)
       - Rank by frequency or importance
       - Identify emerging patterns
       - Use the 'categorize_use_cases' tool
    
    3. **Trends and Patterns** - What patterns emerge from the data?
       - Temporal trends (how usage changed over time)
       - Demographic variations
       - Growth indicators
       - Use the 'analyze_trends' tool
    
    Please provide specific numbers, statistics, and evidence from the document.
    """
    
    research_request = f"""
    {research_query}
    
    Document to Analyze:
    {pdf_content[:50000]}
    """
    
    print(f"✓ Query prepared ({len(research_request)} characters)")
    
    # ========================================================================
    # STEP 4: Run Graph (UNCOMMENT WHEN READY)
    # ========================================================================
    print("\n🔍 Step 4: Running research...")
    print("="*70)
    
    # Import your graph
    try:
        # Replace this with your actual graph import
        from research_graph import graph  # Adjust this import!
        
        print("\n🚀 Executing research graph...\n")
        
        # For synchronous graph.invoke()
        result = graph.invoke(
            {"messages": [{"role": "user", "content": research_request}]},
            config_dict
        )
        
        print("\n" + "="*70)
        print("📄 RESEARCH RESULTS")
        print("="*70)
        print(result)
        
        # OR for async graph.astream()
        import asyncio
        async def run_async():
            async for event in graph.astream(
                {"messages": [{"role": "user", "content": research_request}]},
                config_dict,
                stream_mode="updates"
            ):
                for node_name, node_output in event.items():
                    print(f"\n📍 {node_name}")
                    if "final_report" in node_output:
                        print(node_output["final_report"])
        
        asyncio.run(run_async())
        
    except ImportError as e:
        print(f"⚠️  Graph not imported: {e}")
        print("\n💡 To complete integration:")
        print("   1. Create/import your LangGraph graph")
        print("   2. Update the import line above")
        print("   3. Use config_dict when calling the graph")
        print("\n📦 Your config is ready to use:")
        print(f"   - config_dict available")
        print(f"   - research_request prepared")
        print(f"   - All tools configured")
        
        # Show what would be passed to the graph
        print("\n🔧 Graph would receive:")
        print(f"   Messages: [{'role': 'user', 'content': '{research_query[:50]}...'}]")
        print(f"   Config: {list(config_dict['configurable'].keys())[:5]}...")
    
    except Exception as e:
        print(f"❌ Error running graph: {e}")
        import traceback
        traceback.print_exc()
    
    # ========================================================================
    # STEP 5: Print Summary
    # ========================================================================
    print("\n" + "="*70)
    print("📊 SESSION SUMMARY")
    print("="*70)
    
    config.print_summary()
    
    return config, config_dict, research_request


if __name__ == "__main__":
    config, config_dict, request = setup_and_run_research()
    
    print("\n" + "="*70)
    print("✅ Setup complete! Next steps:")
    print("="*70)
    print("1. If you have a graph, update the import in Step 4")
    print("2. Your config_dict is ready to use")
    print("3. Your research_request is prepared")
    print("\nExample graph usage:")
    print("   result = graph.invoke(")
    print("       {'messages': [{'role': 'user', 'content': request}]},")
    print("       config_dict")
    print("   )")
````
